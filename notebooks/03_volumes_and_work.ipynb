## Mathematical idea

For rotation of $y=f(x)\ge 0$ about the $x$-axis, $V=\pi\int_a^b f(x)^2dx$. For variable force, $W=\int_a^b F(x)dx$.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import quad
from IPython.display import display
import ipywidgets as widgets
plt.rcParams.update({'figure.figsize': (7, 4.5), 'axes.grid': True})
```

## Worked visualization

```python
x=np.linspace(0,1,300);f=np.sqrt(x)
fig,ax=plt.subplots();ax.plot(x,f);ax.fill_between(x,0,f,alpha=.2);ax.set_title(r'Profile $y=\sqrt{x}$ rotated about x-axis');plt.show()
print('Exact volume:',np.pi/2)
```

## Interactive experiment

Move the controls and describe the change. Run the cell in a live Jupyter kernel to display the widgets.

```python
def washers(n=8):
    mids=(np.arange(n)+.5)/n;radius=np.sqrt(mids)
    approx=np.pi*np.sum(radius**2)/n
    fig,ax=plt.subplots();ax.bar(mids,np.pi*radius**2,width=1/n,alpha=.5,label='midpoint slices');xx=np.linspace(0,1,200);ax.plot(xx,np.pi*xx,label='cross-sectional area');ax.legend();plt.show()
    print(f'Approximate volume = {approx:.6f}; exact = {np.pi/2:.6f}')
widgets.interact(washers,n=widgets.IntSlider(value=8,min=2,max=50));
```
