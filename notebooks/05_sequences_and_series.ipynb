## Mathematical idea

A series $\\sum_{n=1}^\\infty a_n$ converges if its partial sums $s_N=\\sum_{n=1}^N a_n$ converge. For $|r|<1$, $\\sum_{n=0}^\\infty ar^n=a/(1-r)$.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import quad
from IPython.display import display
import ipywidgets as widgets
plt.rcParams.update({'figure.figsize': (7, 4.5), 'axes.grid': True})
```

## Worked visualization

```python
N=np.arange(1,101);s=np.cumsum(1/2.**N)
fig,ax=plt.subplots();ax.plot(N,s,'o-',ms=2);ax.axhline(1,ls='--');ax.set_title('Geometric partial sums');plt.show()
```

## Interactive experiment

Move the controls and describe the change. Run the cell in a live Jupyter kernel to display the widgets.

```python
def partial_sums(r=.5,n=20):
    k=np.arange(n);s=np.cumsum(r**k)
    fig,ax=plt.subplots();ax.plot(k+1,s,'o-');
    if abs(r)<1:ax.axhline(1/(1-r),ls='--',label='infinite sum');ax.legend()
    ax.set_xlabel('N');ax.set_ylabel('Partial sum');plt.show()
widgets.interact(partial_sums,r=widgets.FloatSlider(value=.5,min=-1.2,max=1.2,step=.1),n=widgets.IntSlider(value=20,min=3,max=80));
```
