## Mathematical idea

For $\\sqrt{a^2-x^2}$, substitute $x=a\\sin\\theta$. Then $dx=a\\cos\\theta\\,d\\theta$ and $\\sqrt{a^2-x^2}=a\\cos\\theta$ on the usual substitution interval.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import quad
from IPython.display import display
import ipywidgets as widgets
plt.rcParams.update({'figure.figsize': (7, 4.5), 'axes.grid': True})
```

## Worked visualization

```python
a=2.;x=np.linspace(-a,a,400)
fig,ax=plt.subplots();ax.plot(x,np.sqrt(np.maximum(a*a-x*x,0)));ax.set_aspect('equal');ax.set_title(r'$y=\sqrt{a^2-x^2}$');plt.show()
import sympy as sp
z=sp.symbols('x');print('Partial fractions:',sp.apart((z-7)/((z+1)*(z-3)),z))
```

## Interactive experiment

Move the controls and describe the change. Run the cell in a live Jupyter kernel to display the widgets.

```python
def trig_triangle(a=2.,theta=.5):
    x=a*np.sin(theta);y=a*np.cos(theta)
    fig,ax=plt.subplots();ax.plot([0,x,x,0],[0,0,y,0],'-o');ax.set_xlim(0,4);ax.set_ylim(0,4);ax.set_aspect('equal');ax.set_title(f'x={x:.2f}, sqrt(a²-x²)={y:.2f}');plt.show()
widgets.interact(trig_triangle,a=widgets.FloatSlider(value=2,min=1,max=3,step=.25),theta=widgets.FloatSlider(value=.5,min=.05,max=1.45,step=.05));
```
