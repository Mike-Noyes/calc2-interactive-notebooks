## Mathematical idea

The degree-$N$ Taylor polynomial about $a$ is $T_N(x)=\sum_{k=0}^N f^{(k)}(a)(x-a)^k/k!$. For $e^x$ about zero, $T_N(x)=\sum_{k=0}^N x^k/k!$.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import quad
from IPython.display import display
import ipywidgets as widgets
plt.rcParams.update({'figure.figsize': (7, 4.5), 'axes.grid': True})
```

## Worked visualization

```python
from math import factorial
x=np.linspace(-3,3,400);fig,ax=plt.subplots();ax.plot(x,np.exp(x),label='exp(x)');
for N in [1,3,7]:ax.plot(x,sum(x**k/factorial(k) for k in range(N+1)),label=f'N={N}')
ax.legend();plt.show()
```

## Interactive experiment

Move the controls and describe the change. Run the cell in a live Jupyter kernel to display the widgets.

```python
from math import factorial
def taylor_demo(function='sin',N=5):
    x=np.linspace(-5,5,500)
    if function=='exp':exact=np.exp(x);approx=sum(x**k/factorial(k) for k in range(N+1))
    elif function=='sin':exact=np.sin(x);approx=sum((-1)**k*x**(2*k+1)/factorial(2*k+1) for k in range((N+1)//2))
    else:exact=np.cos(x);approx=sum((-1)**k*x**(2*k)/factorial(2*k) for k in range(N//2+1))
    fig,ax=plt.subplots();ax.plot(x,exact,label='function');ax.plot(x,approx,label='polynomial');ax.set_ylim(-5,5);ax.legend();plt.show()
widgets.interact(taylor_demo,function=widgets.Dropdown(options=['sin','cos','exp']),N=widgets.IntSlider(value=5,min=0,max=15));
```
