## Mathematical idea

For $x=x(t),y=y(t)$, $\\frac{dy}{dx}=\\frac{y\'(t)}{x\'(t)}$ where $x\'\\ne0$. For a polar curve $r=f(\\theta)$, $A=\\frac12\\int_\\alpha^\\beta r^2\\,d\\theta$.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import quad
from IPython.display import display
import ipywidgets as widgets
plt.rcParams.update({'figure.figsize': (7, 4.5), 'axes.grid': True})
```

## Worked visualization

```python
t=np.linspace(0,1,250);fig,ax=plt.subplots();ax.plot(t*t,t*t*t);ax.set_title(r'$x=t^2,\ y=t^3$');plt.show()
print('Arc length:',quad(lambda t:np.hypot(2*t,3*t*t),0,1)[0])
```

## Interactive experiment

Move the controls and describe the change. Run the cell in a live Jupyter kernel to display the widgets.

```python
def rose(k=3,phase=0.):
    th=np.linspace(0,2*np.pi,1200);r=np.cos(k*th+phase)
    fig,ax=plt.subplots(subplot_kw={'projection':'polar'});ax.plot(th,r);ax.set_title(f'r = cos({k}θ + {phase:.2f})');plt.show()
widgets.interact(rose,k=widgets.IntSlider(value=3,min=1,max=8),phase=widgets.FloatSlider(value=0,min=0,max=3.14,step=.1));
```
