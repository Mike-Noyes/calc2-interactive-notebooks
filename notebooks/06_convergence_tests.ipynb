## Mathematical idea

The $p$-series $\\sum_{n=1}^\\infty n^{-p}$ converges for $p>1$. For alternating decreasing terms tending to zero, the alternating-series error is bounded by the first omitted term.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import quad
from IPython.display import display
import ipywidgets as widgets
plt.rcParams.update({'figure.figsize': (7, 4.5), 'axes.grid': True})
```

## Worked visualization

```python
n=np.arange(1,1001);fig,ax=plt.subplots();
for p in [0.8,1,1.5,2]:ax.plot(n,np.cumsum(n.astype(float)**(-p)),label=f'p={p}')
ax.legend();ax.set_title('p-series partial sums');plt.show()
```

## Interactive experiment

Move the controls and describe the change. Run the cell in a live Jupyter kernel to display the widgets.

```python
def p_series(p=1.5,n=100):
    k=np.arange(1,n+1,dtype=float);s=np.cumsum(k**(-p));
    fig,ax=plt.subplots();ax.plot(k,s);ax.set_xlabel('N');ax.set_ylabel('Partial sum');plt.show()
    print('Partial sum:',s[-1], 'Integral-test tail estimate:',(n**(1-p)/(p-1) if p>1 else 'not applicable for p<=1'))
widgets.interact(p_series,p=widgets.FloatSlider(value=1.5,min=.5,max=3,step=.1),n=widgets.IntSlider(value=100,min=10,max=1000,step=10));
```
