## Mathematical idea

The geometric power series $\sum_{n=0}^{\infty} x^n=1/(1-x)$ for $|x|<1$. Endpoint behavior must be checked separately for general power series.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import quad
from IPython.display import display
import ipywidgets as widgets
plt.rcParams.update({'figure.figsize': (7, 4.5), 'axes.grid': True})
```

## Worked visualization

```python
x=np.linspace(-.9,.9,350);fig,ax=plt.subplots();ax.plot(x,1/(1-x),label='exact');
for N in [2,5,12]:ax.plot(x,sum(x**k for k in range(N+1)),label=f'N={N}')
ax.legend();plt.show()
```

## Interactive experiment

Move the controls and describe the change. Run the cell in a live Jupyter kernel to display the widgets.

```python
def geom_poly(N=5,xmax=.95):
    x=np.linspace(-xmax,xmax,400);approx=sum(x**k for k in range(N+1));
    fig,ax=plt.subplots();ax.plot(x,1/(1-x),label='exact');ax.plot(x,approx,label=f'degree {N}');ax.legend();plt.show()
widgets.interact(geom_poly,N=widgets.IntSlider(value=5,min=0,max=35),xmax=widgets.FloatSlider(value=.95,min=.2,max=1.3,step=.05));
```
