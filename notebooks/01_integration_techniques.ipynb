## Mathematical idea

Integration by parts: $\\int u\\,dv=uv-\\int v\\,du$. A central example is $\\int x\\ln x\\,dx=\\frac{x^2}{2}\\ln x-\\frac{x^2}{4}+C$.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import quad
from IPython.display import display
import ipywidgets as widgets
plt.rcParams.update({'figure.figsize': (7, 4.5), 'axes.grid': True})
```

## Worked visualization

```python
x=np.linspace(.2,3,300)
f=x*np.log(x)
F=.5*x*x*np.log(x)-.25*x*x
fig,ax=plt.subplots(); ax.plot(x,f,label=r'$x\ln x$'); ax.plot(x,np.gradient(F,x),ls='--',label='numerical derivative of antiderivative'); ax.legend(); plt.show()
print('Integral from 1 to 2:',quad(lambda t:t*np.log(t),1,2)[0])
```

## Interactive experiment

Move the controls and describe the change. Run the cell in a live Jupyter kernel to display the widgets.

```python
def integration_demo(power=1,upper=2.0):
    xs=np.linspace(.1,upper,250); y=xs**power*np.log(xs)
    fig,ax=plt.subplots();ax.plot(xs,y);ax.fill_between(xs,y,alpha=.25);ax.set_title(rf'$\int_{{0.1}}^{{{upper:.1f}}}x^{{{power}}}\ln x\,dx$');plt.show()
    print('Numerical area:',quad(lambda t:t**power*np.log(t),.1,upper)[0])
widgets.interact(integration_demo,power=widgets.IntSlider(value=1,min=0,max=5),upper=widgets.FloatSlider(value=2,min=.5,max=4,step=.1));
```
